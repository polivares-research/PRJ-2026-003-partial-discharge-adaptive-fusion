# V5 gate and scientific verdict

This wrapper reads versioned V5 gate/fusion reports, checks that holdouts were closed during selection, and keeps MATLAB and VSB summaries separate.

```python
import json
from pathlib import Path
path = Path('reports/metrics/v5-pulse-aware/development_gate.json')
if path.exists():
    report = json.loads(path.read_text())
    print(report.get('gate_status'), report.get('selected_candidate'))
else:
    print('V5 gate report is not available yet; no scientific verdict is emitted.')
```
